```python
import pandas as pd
import random

# Constants
num_employees = 50
employees = [f'Person{i+1}' for i in range(num_employees)]
shifts = ['1st Shift', '2nd Shift', '3rd Shift']
hours_map = {
    '1st Shift': '0600 - 1500',
    '2nd Shift': '1200 - 2100',
    '3rd Shift': '1400-2300'
}
departments = ['Customer Service', 'Grocery', 'Meat']

# Data generation
data = []
for person in employees:
    shift = random.choice(shifts)
    hours = hours_map[shift]
    department = random.choice(departments)
    data.append([person, shift, department, hours])

# Create DataFrame
df = pd.DataFrame(data, columns=['Employee', 'Shift', 'Department', 'Hours'])

# Save to CSV
df.to_csv('fake_employee_data.csv', index=False)

print(df.head())  # preview the data
```
